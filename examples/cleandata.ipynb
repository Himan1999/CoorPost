```python

# Read an Excel (.xls or .xlsx) file
df = pd.read_excel("Rastriya Prajatantra Party - राप्रपा _language.xlsx")

# Show first 5 rows
print(df.head())
```

```python
df.info()
```

```python
df["likes"] = df["likes"].replace("", None).fillna(400)
df["shares"] = df["shares"].replace("", None).fillna(257)
```

```python


now = pd.Timestamp.now()
start = now - pd.Timedelta(days=7)

# Generate random valid timestamps as pandas Timestamps
random_times = pd.to_datetime(
    start + (now - start) * np.random.rand(len(df))
)

df["timestamp"] = random_times
```

```python
df.info()
```

```python
# Add incremental unique post_id starting from 1
df["post_id"] = range(1, len(df) + 1)
```

```python
cols = ["post_id"] + [col for col in df.columns if col != "post_id"]
df = df[cols]
```

```python
df = df.rename(columns={
    "post_id": "post_id",
    "user/name": "page_id",
    "url": "shared_url",
    "timestamp": "created_time",
    "likes": "likes",
    "shares": "shares"
})
```

```python
df_selected = df[["post_id", "page_id", "shared_url", "created_time", "likes", "shares"]]
```
